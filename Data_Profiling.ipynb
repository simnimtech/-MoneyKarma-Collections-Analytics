```python
#foreign key validation
import pandas as pd
import os

data_path = "/Users/simrannimje/Desktop/Collections_Analytics_Assignment/Data/Raw/"

# Load datasets
borrowers = pd.read_csv(data_path + "borrowers.csv", dtype=str)
accounts = pd.read_csv(data_path + "accounts.csv", dtype=str)
agents = pd.read_csv(data_path + "agents.csv", dtype=str)
calls = pd.read_csv(data_path + "calls.csv", dtype=str)
payments = pd.read_csv(data_path + "payments.csv", dtype=str)
campaigns = pd.read_csv(data_path + "campaigns.csv", dtype=str)
```

```python
def check_relationship(child_df, parent_df, 
                       child_column, parent_column,
                       relationship_name):
    child_values = set(
        child_df[child_column].dropna()
    )
    parent_values = set(
        parent_df[parent_column].dropna()
    )
    matched = child_values.intersection(parent_values)
    unmatched = child_values - parent_values
    print("\nRelationship:", relationship_name)
    print("Unique IDs in child:", len(child_values))
    print("Matched IDs:", len(matched))
    print("Unmatched IDs:", len(unmatched))
    print("Match Percentage:",
          round(len(matched) / len(child_values) * 100, 2)
          if child_values else 0)
    return unmatched
```

```python
#Accounts --- Borrowers Every account should ideally map to a borrower
#Payments --- Accounts Every payment should map to an account
#Calls --- Accounts Every call should map to an account
#Calls --- Agents Every identified agent should map to an agent record
#Calls --- Campaigns Every call should map to a campaign

# Accounts -> Borrowers
check_relationship(
    accounts,
    borrowers,
    "borrower_id",
    "borrower_id",
    "Accounts to Borrowers"
)

# Payments -> Accounts
check_relationship(
    payments,
    accounts,
    "account_id",
    "account_id",
    "Payments to Accounts"
)

# Calls -> Accounts
check_relationship(
    calls,
    accounts,
    "account_id",
    "account_id",
    "Calls to Accounts"
)

# Calls -> Agents
check_relationship(
    calls,
    agents,
    "agent_id",
    "agent_id",
    "Calls to Agents"
)

# Calls -> Campaigns
check_relationship(
    calls,
    campaigns,
    "campaign_id",
    "campaign_id",
    "Calls to Campaigns"
)
```

```python
#improving foreign key validation
def detailed_fk_check(child_df, parent_df,
                      child_column, parent_column,
                      relationship_name):

    parent_values = set(
        parent_df[parent_column].dropna()
    )

    child_values = child_df[child_column]
    missing = child_values.isna().sum()
    valid = child_values.isin(parent_values).sum()
    invalid = (
        child_values.notna() &
        ~child_values.isin(parent_values)
    ).sum()

    total = len(child_df)
    result = {
        "Relationship": relationship_name,
        "Total Records": total,
        "Valid Records": valid,
        "Missing Foreign Keys": missing,
        "Invalid Foreign Keys": invalid,
        "Valid Match %": round(valid / total * 100, 2)
    }

    return result
```

```python
relationships = [

    (accounts, borrowers,
     "borrower_id", "borrower_id",
     "Accounts → Borrowers"),

    (payments, accounts,
     "account_id", "account_id",
     "Payments → Accounts"),

    (calls, accounts,
     "account_id", "account_id",
     "Calls → Accounts"),

    (calls, agents,
     "agent_id", "agent_id",
     "Calls → Agents"),

    (calls, campaigns,
     "campaign_id", "campaign_id",
     "Calls → Campaigns")
]

fk_results = []

for child, parent, child_col, parent_col, name in relationships:

    result = detailed_fk_check(
        child, parent,
        child_col, parent_col,
        name
    )

    fk_results.append(result)

fk_report = pd.DataFrame(fk_results)

display(fk_report)
```

```python
# Convert date columns
promises_to_pay["promised_date"] = pd.to_datetime(
    promises_to_pay["promised_date"],
    errors="coerce"
)

successful_payments_clean["event_at"] = pd.to_datetime(
    successful_payments_clean["event_at"],
    errors="coerce"
)

# Keep successful payments
payment_check = successful_payments_clean[
    ["account_id", "event_at", "amount"]
].copy()

payment_check["amount"] = pd.to_numeric(
    payment_check["amount"], errors="coerce"
)

# Match PTPs with successful payments on account_id
ptp_payment_detail = promises_to_pay.merge(
    payment_check,
    on="account_id",
    how="left",
    suffixes=("_ptp", "_payment")
)

# Payments made between PTP creation and promised date
ptp_payment_detail["PAYMENT_WITHIN_PROMISE_WINDOW"] = (
    (ptp_payment_detail["event_at_payment"] >= ptp_payment_detail["event_at_ptp"]) &
    (ptp_payment_detail["event_at_payment"] <= ptp_payment_detail["promised_date"])
)

# Summarise matching payments
ptp_payment_summary = (
    ptp_payment_detail
    .groupby("ptp_id")
    .agg(
        status=("status", "first"),
        promised_amount=("promised_amount", "first"),
        matched_payment_count=("PAYMENT_WITHIN_PROMISE_WINDOW", "sum"),
        matched_payment_amount=(
            "amount",
            lambda x: x[
                ptp_payment_detail.loc[x.index, "PAYMENT_WITHIN_PROMISE_WINDOW"]
            ].sum()
        )
    )
    .reset_index()
)

display(ptp_payment_summary.head(10))
```

```python
# Compare matched payment amount with promised amount

ptp_payment_summary["AMOUNT_FULFILLED"] = (
    ptp_payment_summary["matched_payment_amount"] >=
    ptp_payment_summary["promised_amount"]
)

# Summarise by PTP status
ptp_amount_validation = (
    ptp_payment_summary
    .groupby("status")
    .agg(
        total_promises=("ptp_id", "count"),
        promises_with_payment=(
            "matched_payment_count",
            lambda x: (x > 0).sum()
        ),
        promises_fully_paid=("AMOUNT_FULFILLED", "sum")
    )
)

ptp_amount_validation["FULL_PAYMENT_RATE_%"] = (
    ptp_amount_validation["promises_fully_paid"] /
    ptp_amount_validation["total_promises"] * 100
).round(2)

display(ptp_amount_validation)
```

```python
# Convert opened_at to datetime
accounts["opened_at"] = pd.to_datetime(
    accounts["opened_at"], errors="coerce"
)

accounts["opening_month"] = (
    accounts["opened_at"].dt.to_period("M")
)

print("ACCOUNT OPENING DATE RANGE")
print("First:", accounts["opened_at"].min())
print("Last:", accounts["opened_at"].max())

print("\nACCOUNTS OPENED BY MONTH")
display(
    accounts.groupby("opening_month")
    .size()
    .to_frame("accounts_opened")
)

```
